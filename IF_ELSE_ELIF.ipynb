```python
idade = int(input('Digite sua idade: '))
if idade < 16:
    print('Menor de idade')
elif idade >= 16 and idade < 18:
    print('Voto opcional')
elif idade >= 18 and idade < 70:
    print('Voto obrigatório')
else:
  print('Voto opcional')
```

Crie um programa que peça:

Nome

Idade

O programa deve permitir a entrada somente se a pessoa tiver 18 anos ou mais E estiver com ingresso.

O programa deve perguntar:

Você possui ingresso? (sim/não)

Regras

Se:

idade >= 18 E

ingresso == "sim"

Mostrar: Entrada permitida!

Caso contrário: Entrada não permitida.

Dica: utilize if, else, input, int e and.

```python
nome = input('Digite seu nome ')
print(f'Olá {nome}')

ingresso = input('Possui ingresso? sim / não ')
if ingresso == 'sim':
    print('Você tem ingresso! ')
elif ingresso == 'não':
  print('Você não tem ingresso. ')

idade = int(input('Digite sua idade: '))
if idade >=18 and ingresso == 'sim':
  print(f'Tudo certo, {nome} tenha um ótimo evento ')
else:
  print('Entrada não permitida ')
```
